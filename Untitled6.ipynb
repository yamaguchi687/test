```python
import os
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
from PIL import Image, UnidentifiedImageError
import numpy as np
from datetime import datetime
from accunet_model import ACCUNet
from tqdm import tqdm
from torch.utils.tensorboard import SummaryWriter
import torchvision

# === 設定 ===
NUM_CLASSES = 22
BATCH_SIZE = 4
CHECKPOINT_PATH = "/content/drive/MyDrive/accunet_checkpoint.pth"  # Drive上に保存
IMG_DIR = "/content/drive/MyDrive/bdd100k/train/img"
MASK_DIR = "/content/drive/MyDrive/bdd100k/train/masks2"
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# === 高速化設定（Colabでは効果限定） ===
os.environ['OMP_NUM_THREADS'] = str(os.cpu_count())
os.environ['MKL_NUM_THREADS'] = str(os.cpu_count())

# === 画像変換 ===
transform_img = transforms.Compose([
    transforms.Resize((256, 512)),
    transforms.ToTensor()
])

# === Dataset定義（例外処理付き） ===
class BDDDataset(Dataset):
    def __init__(self, img_dir, mask_dir, image_size=(512, 256), transform=None):
        self.img_paths = sorted([
            os.path.join(img_dir, f) for f in os.listdir(img_dir) if f.endswith(".jpg")
        ])
        self.mask_paths = sorted([
            os.path.join(mask_dir, f.replace(".jpg", ".jpg.png")) for f in os.listdir(img_dir) if f.endswith(".jpg")
        ])
        self.image_size = image_size
        self.transform = transform

    def __len__(self):
        return len(self.img_paths)

    def __getitem__(self, idx):
        try:
            img_path = self.img_paths[idx]
            mask_path = self.mask_paths[idx]

            img = Image.open(img_path).convert("RGB")
            mask = Image.open(mask_path)

            if self.image_size:
                img = img.resize(self.image_size)
                mask = mask.resize(self.image_size, Image.NEAREST)

            if self.transform:
                img = self.transform(img)
            else:
                img = torch.from_numpy(np.array(img).transpose(2, 0, 1)).float() / 255.0

            mask = np.array(mask, dtype=np.int64)
            return img, torch.tensor(mask)

        except (UnidentifiedImageError, OSError, IOError) as e:
            print(f"⚠️ 読み込み失敗: {img_path} / {mask_path} - {e}")
            return self.__getitem__((idx + 1) % len(self.img_paths))

# === 学習関数 ===
def train():
    model = ACCUNet(in_channels=3, out_channels=NUM_CLASSES).to(DEVICE)
    optimizer = optim.Adam(model.parameters(), lr=1e-4)
    criterion = nn.CrossEntropyLoss()
    start_epoch = 0

    # --- チェックポイントの読み込み ---
    if os.path.exists(CHECKPOINT_PATH):
        print(f"✅ チェックポイントを読み込みます: {CHECKPOINT_PATH}")
        checkpoint = torch.load(CHECKPOINT_PATH, map_location=DEVICE)
        model.load_state_dict(checkpoint['model_state_dict'])
        optimizer.load_state_dict(checkpoint['optimizer_state_dict'])
        start_epoch = checkpoint['epoch'] + 1
    else:
        print("🆕 新規学習を開始します")

    # --- TensorBoardログ設定 ---
    log_dir = "runs/accunet_" + datetime.now().strftime("%Y%m%d-%H%M%S")
    writer = SummaryWriter(log_dir)

    # --- データ読み込み ---
    train_dataset = BDDDataset(IMG_DIR, MASK_DIR, transform=transform_img)
    train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True, num_workers=0)

    # --- モデル構造を記録（初回のみ） ---
    if start_epoch == 0:
        img_sample, _ = next(iter(train_loader))
        writer.add_graph(model, img_sample.to(DEVICE))

    num_epochs = 2

    for epoch in range(start_epoch, num_epochs):
        model.train()
        total_loss = 0

        for i, (img, mask) in enumerate(tqdm(train_loader, desc=f"Epoch {epoch+1}")):
            img, mask = img.to(DEVICE), mask.to(DEVICE)

            pred = model(img)
            loss = criterion(pred, mask)

            optimizer.zero_grad()
            loss.backward()
            optimizer.step()

            total_loss += loss.item()
            global_step = epoch * len(train_loader) + i
            writer.add_scalar('Loss/step', loss.item(), global_step)

        # --- 損失ログ ---
        avg_loss = total_loss / len(train_loader)
        writer.add_scalar('Loss/epoch', avg_loss, epoch)
        print(f"📘 Epoch {epoch+1}/{num_epochs}, Loss: {avg_loss:.4f}")

        # --- 可視化用画像ログ ---
        pred_map = torch.argmax(torch.softmax(pred, dim=1), dim=1, keepdim=True)
        gt_map = mask.unsqueeze(1)

        img_grid = torchvision.utils.make_grid(img, normalize=True)
        gt_grid = torchvision.utils.make_grid(gt_map.float() / (NUM_CLASSES - 1))
        pred_grid = torchvision.utils.make_grid(pred_map.float() / (NUM_CLASSES - 1))

        writer.add_image('Epoch/Input', img_grid, epoch)
        writer.add_image('Epoch/GroundTruth', gt_grid, epoch)
        writer.add_image('Epoch/Prediction', pred_grid, epoch)

        # --- チェックポイント保存 ---
        torch.save({
            'epoch': epoch,
            'model_state_dict': model.state_dict(),
            'optimizer_state_dict': optimizer.state_dict(),
            'loss': avg_loss,
        }, CHECKPOINT_PATH)
        print(f"💾 チェックポイントを保存しました: {CHECKPOINT_PATH}")

    writer.close()

    # --- 最終モデル保存 ---
    torch.save({
        'model_state_dict': model.state_dict(),
        'class_names': CLASS_NAMES
    }, os.path.join("/content/drive/MyDrive", "accunet_with_labels.pth"))
    print("✅ モデル保存完了: accunet_with_labels.pth")

# === クラスラベル ===
CLASS_NAMES = [
    'road', 'sidewalk', 'building', 'wall', 'fence',
    'pole', 'traffic light', 'traffic sign', 'vegetation',
    'terrain', 'sky', 'person', 'rider', 'car',
    'truck', 'bus', 'train', 'motorcycle', 'bicycle',
    'drivable area', 'lane'
]

if __name__ == "__main__":
    train()
```
